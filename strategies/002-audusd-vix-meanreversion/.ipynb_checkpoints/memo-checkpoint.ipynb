## Abstract

AUD is a structural risk-on currency. VIX measures implied volatility on S&P 500 options and is a proxy for global fear. When VIX spikes sharply in a single day, AUD/USD tends to sell off as risk appetite collapses. This memo tests whether that selloff overshoots -- creating a mean-reversion opportunity -- and whether the edge is consistent across different holding periods and market regimes.

---

## 1. Hypothesis & Mechanism

**Hypothesis:** When VIX rises 8% or more in a single trading day, AUD/USD overshoots to the downside relative to its historical relationship with VIX. Over the following 1-5 trading days, AUD/USD partially reverts, producing a positive return for a long AUD/USD position entered at the close of the spike day.

**Mechanism:** Three structural reasons this edge might exist:

1. **Overshoot/snap-back**: Risk-off flows are driven by momentum and fear, not fundamentals. AUD sells off beyond what the VIX move justifies, then corrects when fear subsides.
2. **VIX mean-reversion**: VIX itself mean-reverts rapidly after spikes. As fear subsides, risk-on currencies like AUD recover.
3. **Carry unwind then re-entry**: AUD is a high-yielding G10 currency. Carry traders unwind on fear spikes but re-enter quickly once volatility stabilizes, creating buying pressure.

**Why it might not persist:** This relationship is well-known. The question is whether the *magnitude* of the overshoot is large enough to trade after transaction costs.

---

## 2. Signal Construction

The signal is simple:

$$\text{VIX Return}_t = \frac{VIX_t - VIX_{t-1}}{VIX_{t-1}}$$

**Entry:** When $\text{VIX Return}_t \geq 0.08$ (8% single-day spike), go **long AUD/USD** at the close.

**Exit:** After $N$ trading days (we test $N = 1, 3, 5$).

**Why long AUD/USD?** We're betting AUD recovers after the overshoot. Long AUD/USD profits when AUD strengthens (i.e. the pair goes up).

**Parameters:** VIX spike threshold = 8%, holding periods tested = 1, 3, 5 days

```python
# ── What this cell does ──────────────────────────────────────────────────────
# Imports the libraries we need and downloads real price data.
# yfinance pulls free data from Yahoo Finance.
# VIX ticker is ^VIX, AUD/USD is AUDUSD=X

import pandas as pd
import numpy as np
import yfinance as yf
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings
warnings.filterwarnings('ignore')

# ── Parameters ───────────────────────────────────────────────────────────────
START          = '2005-01-01'   # enough history to test across multiple regimes
END            = '2024-12-31'
VIX_THRESHOLD  = 0.08           # 8% single-day VIX spike
COST_BPS       = 3              # AUD/USD is liquid, lower cost than KRW
HOLD_DAYS_LIST = [1, 3, 5]      # we test all three

# ── Download data ────────────────────────────────────────────────────────────
vix_raw = yf.download('^VIX',     start=START, end=END, auto_adjust=True)
aud_raw = yf.download('AUDUSD=X', start=START, end=END, auto_adjust=True)

vix = vix_raw[['Close']].copy()
aud = aud_raw[['Close']].copy()
vix.columns = ['vix']
aud.columns = ['aud']

# ── Combine into one dataframe ───────────────────────────────────────────────
# inner join = only keep days where BOTH series have data
df = vix.join(aud, how='inner').dropna()

print(f"Data loaded: {len(df)} trading days")
print(f"Date range:  {df.index[0].date()} → {df.index[-1].date()}")
df.tail()
```

## 3. Exploratory Analysis

Before backtesting, we first look at the raw relationship. If the hypothesis is right, we should see AUD/USD fall on days when VIX spikes, and recover in subsequent days.

```python
# ── What this cell does ──────────────────────────────────────────────────────
# Calculates daily returns for both VIX and AUD/USD.
# Then plots them together so we can see the inverse relationship visually.
# This is the "does the basic premise hold" check before we do any backtesting.

df['vix_ret'] = df['vix'].pct_change()   # VIX daily % change
df['aud_ret'] = df['aud'].pct_change()   # AUD/USD daily % change

# ── Correlation check ────────────────────────────────────────────────────────
corr = df['vix_ret'].corr(df['aud_ret'])
print(f"VIX vs AUD/USD daily return correlation: {corr:.3f}")
print(f"(Negative = inverse relationship, which is what we expect)")

# ── How often does VIX spike 8%+? ────────────────────────────────────────────
spikes = df[df['vix_ret'] >= VIX_THRESHOLD]
print(f"\nVIX spikes 8%+ : {len(spikes)} occurrences over {len(df)} trading days")
print(f"Frequency      : roughly once every {len(df)//len(spikes)} trading days")

# ── What does AUD do on spike days? ─────────────────────────────────────────
print(f"\nAUD/USD return on VIX spike days:")
print(f"  Mean   : {spikes['aud_ret'].mean():.3%}")
print(f"  Median : {spikes['aud_ret'].median():.3%}")
print(f"  % negative: {(spikes['aud_ret'] < 0).mean():.1%}")
```

```python
# ── What this cell does ──────────────────────────────────────────────────────
# Scatter plot: each dot is one trading day.
# X axis = VIX return that day, Y axis = AUD/USD return that day.
# Red dots = the 8%+ spike days we're going to trade.
# If the hypothesis is right, red dots should cluster in bottom-right
# (VIX up a lot, AUD down a lot = potential overshoot to fade)

normal_days = df[df['vix_ret'] < VIX_THRESHOLD].dropna()
spike_days  = df[df['vix_ret'] >= VIX_THRESHOLD].dropna()

fig = go.Figure()
fig.add_trace(go.Scatter(
    x=normal_days['vix_ret'] * 100,
    y=normal_days['aud_ret'] * 100,
    mode='markers',
    name='Normal days',
    marker=dict(color='#aaaaaa', size=4, opacity=0.4)
))
fig.add_trace(go.Scatter(
    x=spike_days['vix_ret'] * 100,
    y=spike_days['aud_ret'] * 100,
    mode='markers',
    name='VIX spike days (8%+)',
    marker=dict(color='#e74c3c', size=8),
    text=spike_days.index.strftime('%Y-%m-%d'),
    hovertemplate='%{text}<br>VIX: +%{x:.1f}%<br>AUD: %{y:.2f}%'
))
fig.add_hline(y=0, line_dash='dash', line_color='black', opacity=0.3)
fig.add_vline(x=8, line_dash='dash', line_color='#e74c3c', opacity=0.5)
fig.update_layout(
    title='VIX vs AUD/USD Daily Returns | 2005–2024',
    xaxis_title='VIX Daily Return (%)',
    yaxis_title='AUD/USD Daily Return (%)',
    template='plotly_white', height=500
)
fig.show()
```

## 4. Event Study: What Happens to AUD After VIX Spikes?

Before a full backtest, we run an event study. For every VIX spike day, we look at what AUD/USD does over the next 10 trading days. If mean-reversion is real, we should see negative AUD returns on day 0 followed by positive returns on days 1-5.

```python
# ── What this cell does ──────────────────────────────────────────────────────
# Event study: for each VIX spike, we record AUD/USD returns for
# the 10 days AFTER the spike. Then we average across all events.
# This tells us: on average, how does AUD behave in the days after a VIX spike?
# This is the core question of the hypothesis.

WINDOW = 10   # days to look forward
event_returns = []

spike_indices = df.index[df['vix_ret'] >= VIX_THRESHOLD]

for spike_date in spike_indices:
    loc = df.index.get_loc(spike_date)
    if loc + WINDOW >= len(df):
        continue
    # Get AUD/USD returns for the next WINDOW days
    fwd_rets = df['aud_ret'].iloc[loc+1 : loc+WINDOW+1].values
    event_returns.append(fwd_rets)

event_matrix = np.array(event_returns)
avg_fwd_rets  = event_matrix.mean(axis=0)
cum_fwd_rets  = np.cumprod(1 + avg_fwd_rets) - 1

# ── Plot cumulative average return after spike ───────────────────────────────
fig = go.Figure()
fig.add_trace(go.Bar(
    x=list(range(1, WINDOW+1)),
    y=avg_fwd_rets * 100,
    name='Avg Daily Return',
    marker_color=['#27ae60' if r > 0 else '#e74c3c' for r in avg_fwd_rets]
))
fig.add_trace(go.Scatter(
    x=list(range(1, WINDOW+1)),
    y=cum_fwd_rets * 100,
    name='Cumulative Return',
    line=dict(color='#1a1a2e', width=2),
    yaxis='y2'
))
fig.update_layout(
    title=f'AUD/USD Average Return in {WINDOW} Days After VIX Spike (8%+) | {len(spike_indices)} events',
    xaxis_title='Days After VIX Spike',
    yaxis_title='Avg Daily Return (%)',
    yaxis2=dict(title='Cumulative Return (%)', overlaying='y', side='right'),
    template='plotly_white', height=450,
    legend=dict(orientation='h', y=-0.15)
)
fig.show()

print("Average cumulative AUD/USD return after VIX spike:")
for d in [1, 3, 5, 10]:
    print(f"  Day {d:2d}: {np.cumprod(1 + avg_fwd_rets[:d])[-1] - 1:.3%}")
```

## 5. Backtest Across Holding Periods

We now run the full backtest for each holding period (1, 3, and 5 days). Each trade: enter long AUD/USD at the close of the VIX spike day, exit after N days.

```python
# ── What this cell does ──────────────────────────────────────────────────────
# Full backtest for each holding period.
# For every VIX spike day:
#   - Enter long AUD/USD at close
#   - Hold for N days
#   - Record the return net of transaction costs
# We do this for N = 1, 3, and 5 days.
# No overlapping trades -- if a new spike happens while we're in a trade, skip it.

def run_backtest(df, hold_days, threshold=VIX_THRESHOLD, cost_bps=COST_BPS):
    trades = []
    i = 1
    while i < len(df) - hold_days:
        if df['vix_ret'].iloc[i] >= threshold:
            entry_price = df['aud'].iloc[i + 1]  # next day open approximation
            entry_date  = df.index[i]
            exit_idx    = min(i + hold_days, len(df) - 1)
            exit_price  = df['aud'].iloc[exit_idx]
            exit_date   = df.index[exit_idx]

            raw_ret = -1 * (exit_price - entry_price) / entry_price
            net_ret = raw_ret - (cost_bps / 10000)

            trades.append({
                'entry_date':  entry_date,
                'exit_date':   exit_date,
                'entry_price': entry_price,
                'exit_price':  exit_price,
                'raw_ret':     raw_ret,
                'net_ret':     net_ret,
                'vix_spike':   df['vix_ret'].iloc[i]
            })
            i = exit_idx + 1   # no overlapping trades
        else:
            i += 1
    return pd.DataFrame(trades)

results = {}
for h in HOLD_DAYS_LIST:
    results[h] = run_backtest(df, h)
    print(f"Hold {h}d: {len(results[h])} trades")
```

## 6. Performance Statistics

```python
# ── What this cell does ──────────────────────────────────────────────────────
# Calculates the key performance numbers for each holding period.
# Sharpe ratio: return per unit of risk. Above 1.0 is decent, above 1.5 is good.
# Win rate: % of trades that made money.
# Max drawdown: worst peak-to-trough loss. Tells you how painful the worst period was.
# We compare all three holding periods side by side.

def calc_stats(trades_df, hold_days):
    r = trades_df['net_ret']
    cum = (1 + r).cumprod()
    dd  = (cum - cum.cummax()) / cum.cummax()
    annualization = np.sqrt(252 / hold_days)
    sharpe = (r.mean() / r.std()) * annualization if r.std() != 0 else 0
    sortino_std = r[r < 0].std()
    sortino = (r.mean() / sortino_std) * annualization if sortino_std != 0 else 0
    return {
        'Hold Period':     f'{hold_days}d',
        'Trades':          len(r),
        'Win Rate':        f'{(r > 0).mean():.1%}',
        'Avg Return':      f'{r.mean():.3%}',
        'Total Return':    f'{(1+r).prod()-1:.2%}',
        'Sharpe':          f'{sharpe:.2f}',
        'Sortino':         f'{sortino:.2f}',
        'Max Drawdown':    f'{dd.min():.2%}',
    }

stats_rows = [calc_stats(results[h], h) for h in HOLD_DAYS_LIST]
stats_df = pd.DataFrame(stats_rows)
print(stats_df.to_string(index=False))
```

## 7. Equity Curves

```python
# ── What this cell does ──────────────────────────────────────────────────────
# Plots the equity curve for each holding period on the same chart.
# A good equity curve goes up and to the right, steadily, without huge drops.
# We're looking for which holding period has the smoothest, most consistent curve.

colors = {'1d': '#1a1a2e', '3d': '#4a90d9', '5d': '#27ae60'}

fig = go.Figure()
for h in HOLD_DAYS_LIST:
    t   = results[h]
    cum = (1 + t['net_ret']).cumprod() - 1
    fig.add_trace(go.Scatter(
        x=t['exit_date'],
        y=cum * 100,
        name=f'{h}-day hold',
        line=dict(width=2, color=colors[f'{h}d'])
    ))

fig.update_layout(
    title='Strategy 002: AUD/USD VIX Spike Mean-Reversion | Equity Curves by Hold Period',
    xaxis_title='Date',
    yaxis_title='Cumulative Return (%)',
    template='plotly_white',
    height=500,
    legend=dict(orientation='h', y=-0.15)
)
fig.show()
```

## 8. Regime Analysis

Does the edge hold across different market environments?

```python
# ── What this cell does ──────────────────────────────────────────────────────
# Splits the backtest into 4 macro regimes and checks performance in each.
# A real edge should work across multiple regimes, not just one.
# If it only works in one regime (e.g. post-COVID), it's probably not a real edge.

# Use the 3-day hold as the primary for regime analysis
primary = results[3].copy()
primary['year'] = pd.to_datetime(primary['entry_date']).dt.year

regimes = [
    ('Pre-GFC & Recovery (2005–2012)', (2005, 2012)),
    ('Low-vol bull market (2013–2019)', (2013, 2019)),
    ('COVID + Fed tightening (2020–2022)', (2020, 2022)),
    ('Post-normalization (2023–2024)',  (2023, 2024)),
]

regime_rows = []
for label, (y1, y2) in regimes:
    subset = primary[(primary['year'] >= y1) & (primary['year'] <= y2)]
    if len(subset) == 0:
        continue
    r = subset['net_ret']
    regime_rows.append({
        'Regime':       label,
        'Trades':       len(r),
        'Win Rate':     f'{(r > 0).mean():.1%}',
        'Avg Return':   f'{r.mean():.3%}',
        'Total Return': f'{(1+r).prod()-1:.2%}',
    })

regime_df = pd.DataFrame(regime_rows)
print(regime_df.to_string(index=False))
```

## 9. Threshold Sensitivity

Is 8% the right threshold or did we get lucky picking it? We test 5%, 8%, 10%, 15%, and 20% thresholds.

```python
# ── What this cell does ──────────────────────────────────────────────────────
# Tests multiple VIX spike thresholds to see if 8% is special
# or if the edge exists across a range of thresholds.
# If only 8% works and everything else fails, we probably just got lucky
# with the number -- that's called overfitting.
# If the edge degrades gracefully as threshold increases, it's more credible.

thresholds = [0.05, 0.08, 0.10, 0.15, 0.20]
sensitivity_rows = []

for t in thresholds:
    bt = run_backtest(df, hold_days=3, threshold=t)
    if len(bt) == 0:
        continue
    r = bt['net_ret']
    sharpe = (r.mean() / r.std()) * np.sqrt(252/3) if r.std() != 0 else 0
    sensitivity_rows.append({
        'VIX Threshold': f'{t:.0%}',
        'Trades':        len(bt),
        'Win Rate':      f'{(r > 0).mean():.1%}',
        'Avg Return':    f'{r.mean():.3%}',
        'Sharpe':        f'{sharpe:.2f}',
        'Total Return':  f'{(1+r).prod()-1:.2%}',
    })

sens_df = pd.DataFrame(sensitivity_rows)
print(sens_df.to_string(index=False))
```

```python
## 10. Limitations

1. **Daily data only.** AUD/USD is a 24-hour market. Using daily closes misses intraday entry precision -- the optimal entry might be hours after the VIX spike, not at the close.

2. **Transaction costs are estimated.** AUD/USD is highly liquid with tight spreads, but costs vary significantly by broker, account size, and market conditions during high-volatility events.

3. **No position sizing.** Equal-sized positions on every trade ignores the fact that some VIX spikes are more extreme than others. A volatility-scaled position size (larger position when the spike is bigger) is a natural extension.

4. **Threshold sensitivity risk.** If results are only strong at exactly 8% but not at 7% or 9%, the edge may be an artifact of parameter selection. The sensitivity analysis in Section 9 tests this directly.

5. **Tail risk.** This strategy is structurally short volatility -- it bets on mean-reversion after fear spikes. In a genuine regime shift (2008 GFC, March 2020 COVID crash), VIX does not mean-revert quickly and AUD can fall continuously for weeks. Position sizing and stop-losses are essential for live trading.

6. **Survivorship and look-ahead bias.** We know 2005-2024 was a period that included recovery from multiple crises. A strategy that buys risk-on assets after fear spikes benefits structurally from the fact that markets recovered every time. Future crises may not resolve as quickly.

---

## 11. Conclusion

The AUD/USD VIX spike mean-reversion hypothesis is structurally motivated by three mechanisms: overshoot/snap-back dynamics, VIX mean-reversion, and carry trade re-entry. The event study and backtest results determine whether these mechanisms are large enough to trade after transaction costs. The most important robustness checks are: (1) does the edge hold across regimes, and (2) does performance degrade gracefully as the VIX threshold changes, or is it knife-edge sensitive to the 8% parameter?

---
*Ouro Quant Research | github.com/incongnitoperson-cpu/quant-research*
# ── Out-of-sample test ───────────────────────────────────────────────────────
# Split: 2005-2018 = in-sample (we "discovered" the strategy here)
#        2019-2024 = out-of-sample (we never touched this data)
# If the edge is real, it should hold on data the strategy never saw.
# If it collapses, we got lucky on the in-sample period.

df_in  = df[df.index.year <= 2018]
df_out = df[df.index.year >= 2019]

def quick_stats(trades_df, label):
    r = trades_df['net_ret']
    sharpe = (r.mean() / r.std()) * np.sqrt(252) if r.std() != 0 else 0
    print(f"{label}")
    print(f"  Trades      : {len(r)}")
    print(f"  Win Rate    : {(r > 0).mean():.1%}")
    print(f"  Avg Return  : {r.mean():.3%}")
    print(f"  Total Return: {(1+r).prod()-1:.2%}")
    print(f"  Sharpe      : {sharpe:.2f}")
    print()

for h in [1, 3, 5]:
    in_bt  = run_backtest(df_in,  hold_days=h)
    out_bt = run_backtest(df_out, hold_days=h)
    quick_stats(in_bt,  f"IN-SAMPLE  2005-2018 | {h}d hold")
    quick_stats(out_bt, f"OUT-OF-SAMPLE 2019-2024 | {h}d hold")
    print("─" * 45)
```
